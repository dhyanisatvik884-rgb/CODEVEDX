# Exploratory Data Analysis (EDA) using Python

## Overview
In this section, we build upon our clean dataset to perform Exploratory Data Analysis (EDA). The objective is to look deeper into the data to discover hidden patterns, check statistical distributions, and analyze relationships between different variables.The following libraries are used for the analysis and advanced visualizations:

* Numpy 
* Pandas
* Seaborn

<br>

<br>

**1. Importing our desired libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

<br>

**2. Loading the dataset of our csv file**

```python
data = pd.read_csv("dataset.csv")
df = pd.DataFrame(data)
print(df)
```

<br>

**3. Checking the summary of the dataset to the shape, find the data-type and non-null values in particular column**

```python
print("Shape(Rows, Columns) =", df.shape)
print("\n")
df.info()
```

<br>

**4. Handling Null Values (Replacing null values of 'Age', 'Salary' & 'Experience' by median value and of 'Department' & 'City' by word "UNKNOWN")**

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["Experience"] = df["Experience"].fillna(df["Experience"].median())
df["Department"] = df["Department"].fillna("UNKNOWN")
df["City"] = df["City"].fillna("UNKNOWN")
print(df)
```

<br>

**5. Checking the statistical summary of the columns having numerical values**

```python
df.describe()
```

<br>

**6. Checking whether there is any outlier using box plot**

```python
sns.boxplot(df["Salary"])
```

```python
sns.boxplot(df["Age"])
```

```python
sns.boxplot(df["Experience"])
```

<br>

**7. Bivariate Analysis(Experience vs Salary), using ScatterPlot**

```python
sns.scatterplot(x=df["Experience"], y=df["Salary"])
```

<br>

**8. Bivariate Analysis(Department vs Salary), using Barplot**

```python
sns.barplot(x=df["Department"], y=df["Salary"])
```

<br>

**Multivariate Analysis using heatmap**

```python
ASE = df[['Age', 'Salary', 'Experience']]
sns.heatmap(ASE.corr(), annot=True, cmap='coolwarm')
```
