**1. Loading a CSV File and reading its data**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("dataset.csv")
print(df)
```

<br>

**2. Determining dimensions of dataset (Rows, Columns)**

```python
df.shape
```

<br>

**3. Checking the summary of the dataset to find the data-type and non-null values in particular column**

```python
df.info()
```

<br>

**4. Counting total null values in each column**

```python
df.isnull().sum()
```

<br>

**5. Handling Null Values (Replacing null values of 'Age', 'Salary' & 'Experience' by mean value and of 'Department' & 'City' by word "UNKNOWN")**

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Experience"] = df["Experience"].fillna(df["Experience"].mean())
df["Department"] = df["Department"].fillna("UNKNOWN")
df["City"] = df["City"].fillna("UNKNOWN")
print(df)
```

<br>

**6. A simple Piechart representation of percentage of employees per department**

```python

department_counts = df["Department"].value_counts()
people = department_counts.values
departments = department_counts.index

plt.figure(figsize=(8, 8))
plt.pie(people, labels=departments, autopct="%1.1f%%")

plt.title("Distribution of Employees by Department", fontsize=14, fontweight="bold")
plt.show()
```
