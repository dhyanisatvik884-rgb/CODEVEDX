# Data Visualization with Python

## Overview
In this section, we have focused on the presentation of data insights in visual form so that it becomes easier to understand & communicate with the data. The following python libraries have been used for the same
* Pandas
* Matplotlib
* Seaborn

<br>

<br>

**1. Loading the desired python libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

<br>

**2. Loading our csv file to read its dataset**

```python
data = pd.read_csv("dataset.csv")
df = pd.DataFrame(data)
print(df.head())
```

<br>

**3. Cleaning the dataset by handling its null values(Replacing null values of Age, Salary & Experience by Median value and that of Department and City by "UNKNOWN")**

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["Experience"] = df["Experience"].fillna(df["Experience"].median())
df["Department"] = df["Department"].fillna("UNKNOWN")
df["City"] = df["City"].fillna("UNKNOWN")
print(df.head())
```

<br>

**4. Setting the global chart theme**

```python
sns.set_style("whitegrid")
sns.set_context("talk")
```

<br>

**5. Analyzing the distribution of salary, using histogram**

```python
sns.histplot(df["Salary"])
plt.title("Salary Distribution")
plt.xlabel("Annual Salary")
plt.ylabel("Employees")
plt.show()
```

```python
sns.histplot(df["Salary"], kde=True)
plt.title("Distribution")
plt.xlabel("Salary")
plt.ylabel("Employees")
plt.show()
```

<br>

**6. Outlier Analysis using Box Plot**

```python
sns.boxplot(data=df, x="Salary")
plt.title("Salary Distribution and Outlier Detection")
plt.xlabel("Annual Salary")
plt.show()
```

<br>

**7. Salary VS Experience Relationship**

```python
sns.scatterplot(x=df["Experience"], y=df["Salary"])
plt.title("Impact of Professional Experience on Annual Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Annual Salary")
plt.show()
```

<br>

**8. Multivariate Correlation Matrix**

```python
data = df[['Age', 'Salary', 'Experience']]
mx = data.corr()
sns.heatmap(mx, annot=True, cmap='coolwarm')
plt.title("Correlation Analysis of Employee Metrics")
plt.show()
```
