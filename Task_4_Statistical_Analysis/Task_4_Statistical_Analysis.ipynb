# Statistical Analysis using Python

## Overview
In this section we have concentrated on the use of mathematical and statistical techniques for the interpretation of data, the validation of underlying assumptions and the testing of hypotheses so that we can reach reliable, empirical conclusions from the data set rather than just calculating numbers. The following python libraries have been used for the same
* Pandas
* Scipy

<br>

<br>

**1. Importing the desired libraries**

```python
import pandas as pd
from scipy import stats
```

<br>

**2. Loading our csv file and reading it's dataset as well as handling it's null values**

```python
data = pd.read_csv("dataset.csv")
df = pd.DataFrame(data)
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["Experience"] = df["Experience"].fillna(df["Experience"].median())
df["Department"] = df["Department"].fillna("UNKNOWN")
df["City"] = df["City"].fillna("UNKNOWN")
print(df.head())
```

<br>

**3. Testing the hypothesis between department and salary**

```python
IT_sal = df[df["Department"] == 'IT']['Salary']
HR_sal = df[df["Department"] == 'HR']['Salary']
t_stat, p_val = stats.ttest_ind(IT_sal, HR_sal)
print("t-Statistic(Measure of Distance):", t_stat)
if p_val < 0.05:
    print(f"p-Value(Measure of Probability):, {p_val},\n Reject the Null Hypothesis (H₀)")
else:
    print(f"p-Value(Measure of Probability):, {p_val},\n Fail to Reject the Null Hypothesis (H₀)")
```

<br>

**4. Correlation Significance Test between Experience & Salary using Pearson correlation**

```python
corr_coef, p_value_corr = stats.pearsonr(df['Experience'], df['Salary'])
print("Pearson Correlation Coefficient:",corr_coef)
if p_value_corr < 0.05:
    print(f"p-Value(Measure of Probability):, {p_value_corr},\n Statistically significant relationship between Experience and Salary exists.")
else:
    print(f"p-Value(Measure of Probability):, {p_value_corr},\n No statistically significant relationship between Experience and Salary exists.")
```
