# Reporting and Insight Generation 

## Overview
In this section, we have focused on the presentation of data insights in visual and structured summary form so that it becomes easier to understand & communicate with the data. The following python libraries have been used for the same:
* Pandas

<br>

<br>

**1. Importing the desired libraries**

```python
import pandas as pd
```

<br>

**2. Loading our csv file and reading it's dataset as well as handling it's null values**

```python
data = pd.read_csv("dataset.csv")
df = pd.DataFrame(data)
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["Experience"] = df["Experience"].fillna(df["Experience"].median())
df["Department"] = df["Department"].fillna("UNKNOWN")
df["City"] = df["City"].fillna("UNKNOWN")
print(df.head())
```

<br>

**3. Generating Summary Report**

```python
report = {
    "Total Employees": len(df),
    "Average Age": round(df["Age"].mean(), 2),
    "Average Salary": round(df["Salary"].mean(), 2),
    "Average Experience": round(df["Experience"].mean(), 2),
    "Highest Salary": df["Salary"].max(),
    "Lowest Salary": df["Salary"].min(),
    "Most Common Department": df["Department"].mode()[0],
    "Most Common City": df["City"].mode()[0]
}

for key, value in report.items():
    print(f"{key}: {value}")
```

<br>

**4. Generating Department-wise Report**

```python
department_report = df.groupby("Department").agg(
    Total_Employees=("EmployeeID", "count"),
    Average_Age=("Age", "mean"),
    Average_Salary=("Salary", "mean"),
    Average_Experience=("Experience", "mean")
)

department_report = department_report.round(2)

print(department_report)
```

<br>

**5. City-wise Report**

```python
city_report = df.groupby("City").agg(
    Total_Employees=("EmployeeID", "count"),
    Average_Age=("Age", "mean"),
    Average_Salary=("Salary", "mean"),
    Average_Experience=("Experience", "mean")
)

city_report = city_report.round(2)

print(city_report)
```

<br>

**6. Top 5 Highest Paid Employees**

```python
top_salary = df.nlargest(5, "Salary")[["EmployeeID", "Name", "Department", "Salary", "Experience", "City"]]
print(top_salary)
```

<br>

**7. Top 5 Most Experienced Employees**

```python
top_experience = df.nlargest(5, "Experience")[["EmployeeID", "Name", "Department", "Experience", "Salary", "City"]]
print(top_experience)
```

<br>

**8. Department-wise Salary Ranking**

```python
salary_ranking = df.groupby("Department")["Salary"].mean().sort_values(ascending=False).round(2)
print(salary_ranking)
```

<br>

**9. City-wise Salary Ranking**

```python
city_salary = df.groupby("City")["Salary"].mean().sort_values(ascending=False).round(2)
print(city_salary)
```

<br>

**10. Final Report**

```python
final_report = {
    "Total Employees": len(df),
    "Departments": df["Department"].nunique(),
    "Cities": df["City"].nunique(),
    "Average Salary": round(df["Salary"].mean(), 2),
    "Average Experience": round(df["Experience"].mean(), 2)
}

print(pd.DataFrame(final_report.items(), columns=["Report", "Value"]))
```
